```python
# Task 1: NumPy Fundamentals

**Internship:** InternNova - Week 2

**Topic:** Python for AI & Data Handling

**Marks:** 20

## Introduction

NumPy stands for Numerical Python. It is a Python library used for numerical computations and working with arrays.
```

```python
import numpy as np

# 1. Creating NumPy arrays
arr1 = np.array([10, 20, 30, 40, 50])
print("1D Array:")
print(arr1)

# 2. Creating a 2D array
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("\n2D Array:")
print(arr2)

# 3. Array indexing
print("\nIndexing:")
print("First element:", arr1[0])
print("Second row, third column:", arr2[1, 2])

# 4. Array slicing
print("\nSlicing:")
print(arr1[1:4])

# 5. Shape and dimensions
print("\nShape and Dimensions:")
print("1D Shape:", arr1.shape)
print("1D Dimensions:", arr1.ndim)
print("2D Shape:", arr2.shape)
print("2D Dimensions:", arr2.ndim)

# 6. Mathematical operations
print("\nMathematical Operations:")
print("Addition:", arr1 + 5)
print("Subtraction:", arr1 - 5)
print("Multiplication:", arr1 * 2)
print("Division:", arr1 / 2)

# 7. Statistical operations
print("\nStatistical Operations:")
print("Mean:", np.mean(arr1))
print("Maximum:", np.max(arr1))
print("Minimum:", np.min(arr1))
print("Sum:", np.sum(arr1))
```
