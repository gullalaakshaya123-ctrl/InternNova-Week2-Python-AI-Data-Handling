```python

# Task 5: Student Performance Data Analysis
# InternNova - Week 2

import pandas as pd
import matplotlib.pyplot as plt

# 1. Load dataset
df = pd.read_csv("../Datasets/uncleaned_student_data.csv")

print("ORIGINAL DATASET")
print(df)

# 2. Dataset information
print("\nDataset Information:")
df.info()

print("\nImportant Columns:")
print(df[["Name", "Marks", "Attendance"]])

# 3. Data Cleaning
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# Remove duplicate records first
df = df.drop_duplicates()

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Attendance"] = df["Attendance"].fillna(
    df["Attendance"].mean()
)

print("\nCleaned Dataset:")
print(df)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

# Save cleaned dataset
df.to_csv(
    "mini_project_cleaned_data.csv",
    index=False
)

# 4. Data Analysis
average_marks = df["Marks"].mean()
highest_marks = df["Marks"].max()
lowest_marks = df["Marks"].min()

passed = (df["Marks"] >= 50).sum()
failed = (df["Marks"] < 50).sum()

print("\nSTUDENT PERFORMANCE RESULTS")
print("Average Marks:", round(average_marks, 2))
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
print("Passed Students:", passed)
print("Failed Students:", failed)

# 5. Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(df["Name"], df["Marks"])
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Student Performance")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 6. Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Marks"], bins=5, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.show()

# 7. Pie Chart
plt.figure(figsize=(6, 6))
plt.pie(
    [passed, failed],
    labels=["Passed", "Failed"],
    autopct="%1.1f%%"
)
plt.title("Pass and Fail Analysis")
plt.show()

print("\nMini Project Completed Successfully!")
```
